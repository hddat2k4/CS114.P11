# Import Library

```python
import os
import csv
import pandas as pd
from collections import defaultdict
```

```python
# Mount Drive
from google.colab import drive
drive.mount('/content/drive')
```

```python
# Đường dẫn tới thư mục data
#fol = "data"
fol = "/content/drive/MyDrive/Public"

# Các định dạng được hỗ trợ
valid_extensions = {".jpg", ".jpeg", ".png"}


mssv = defaultdict(float)
sv_brand_counts = defaultdict(lambda: defaultdict(float))

# Duyệt qua từng folder
for _, _, files in os.walk(fol):
    # Duyệt qua từng file
    for file in files:
        #print(file)
        if os.path.splitext(file)[1].lower() in valid_extensions:
            parts = file.split(".") # Tách theo dấu . (22520211-22521070.Honda.1.jpg)
            if len(parts)==4 and parts[3]:
                brand = parts[1]
                sv = parts[0].split('-')
                ctb = 1/len(sv)
                for idx in sv:
                    mssv[idx]+=ctb
                    sv_brand_counts[idx][brand] +=ctb
```

# Export CSV

```python
# Lưu vô file csv
with open("CarDataset-1.csv", "w", newline="") as csvfile:
    writer = csv.writer(csvfile)
    writer.writerow(["MSSV", "All", "Count"])
    for mssv, total_count in mssv.items():
        writer.writerow([mssv, "All", round(total_count, 2)])

with open("CarDataset-2.csv", "w", newline="") as csvfile:
    writer = csv.writer(csvfile)
    writer.writerow(["MSSV", "Brand", "Count"])
    for mssv, brands in sv_brand_counts.items():
        for brand, count in brands.items():
            writer.writerow([mssv, brand, round(count, 2)])
```
